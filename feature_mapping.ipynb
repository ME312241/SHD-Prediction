```python
import csv
from pathlib import Path

RES_FEATURES = [
    "in.hvac_cooling_type",
    "in.energystar_climate_zone_2023",
    "in.misc_pool_heater",
    "in.overhangs",
    "in.geometry_space_combination",
    "in.utility_bill_fuel_oil_fixed_charges",
    "in.geometry_building_number_units_mf",
    "in.ceiling_fan",
    "in.hvac_heating_efficiency",
    "in.tenure",
    "in.water_heater_efficiency",
    "in.utility_bill_natural_gas_fixed_charges",
    "in.duct_location",
    "in.hvac_cooling_partial_space_conditioning",
    "in.misc_gas_grill",
    "in.simulation_control_run_period_end_month",
    "in.insulation_slab",
    "in.lighting_interior_use",
    "in.vacancy_status",
    "in.insulation_ceiling",
    "in.electric_vehicle",
    "out.params.floor_area_attic_ft_2",
    "in.federal_poverty_level",
    "in.area_median_income",
    "in.hvac_secondary_heating_efficiency",
    "in.heating_setpoint_offset_magnitude",
    "in.hvac_cooling_efficiency",
    "in.census_region",
    "in.utility_bill_propane_fixed_charges",
    "in.hvac_has_ducts",
    "in.insulation_wall",
    "in.utility_bill_electricity_marginal_rates",
    "in.insulation_foundation_wall",
    "in.ahs_region",
    "in.emissions_electricity_folders",
    "in.emissions_scenario_names",
    "in.geometry_building_type_height",
    "in.range_spot_vent_hour",
    "in.neighbors",
    "in.cec_climate_zone",
    "in.clothes_washer_presence",
    "in.vintage_acs",
    "in.geometry_building_level_mf",
    "in.insulation_floor",
    "in.hvac_has_zonal_electric_heating",
    "in.occupants",
    "in.bathroom_spot_vent_hour",
    "out.params.flow_rate_mechanical_ventilation_cfm",
    "out.params.floor_area_attic_insulation_increase_ft_2_delta_r_value",
    "in.emissions_fossil_fuel_units",
    "in.doors",
    "in.cooling_setpoint_has_offset",
    "in.emissions_electricity_values_or_filepaths",
    "in.iso_rto_region",
    "in.cooling_setpoint_offset_period",
    "in.income_recs_2015",
    "in.roof_material",
    "in.units_represented",
    "in.geometry_floor_area_bin",
    "out.params.size_heating_system_secondary_k_btu_h",
    "in.heating_setpoint_offset_period",
    "in.representative_income",
    "in.cooking_range_usage_level",
    "in.census_division",
    "in.emissions_types",
    "in.reeds_balancing_area",
    "in.simulation_control_timestep",
    "in.geometry_foundation_type",
    "in.utility_bill_fuel_oil_marginal_rates",
    "out.params.floor_area_foundation_ft_2",
    "in.misc_extra_refrigerator",
    "in.hvac_heating_type",
    "in.geometry_floor_area",
    "in.weather_file_city",
    "in.sqft",
    "in.interior_shading",
    "in.ashrae_iecc_climate_zone_2004",
    "in.misc_hot_tub_spa",
    "in.geometry_building_type_recs",
    "in.clothes_washer",
    "in.usage_level",
    "in.battery",
    "in.clothes_dryer_usage_level",
    "in.door_area",
    "in.income_recs_2020",
    "in.windows",
    "out.params.duct_unconditioned_surface_area_ft_2",
    "in.geometry_building_horizontal_location_sfa",
    "in.hvac_shared_efficiencies",
    "in.dehumidifier",
    "in.plug_load_diversity",
    "in.heating_fuel",
    "in.clothes_dryer",
    "in.geometry_building_horizontal_location_mf",
    "in.emissions_electricity_units",
    "in.geometry_wall_type",
    "in.dishwasher",
    "in.has_pv",
    "out.params.floor_area_lighting_ft_2",
    "in.corridor",
    "in.household_has_tribal_persons",
    "in.hvac_system_single_speed_ashp_charge",
    "in.cooking_range",
    "in.geometry_wall_exterior_finish",
    "in.hvac_heating_type_and_fuel",
    "in.hvac_system_single_speed_ac_charge",
    "in.holiday_lighting",
    "in.building_america_climate_zone",
    "in.puma_metro_status",
    "in.utility_bill_electricity_fixed_charges",
    "in.utility_bill_natural_gas_marginal_rates",
    "in.insulation_rim_joist",
    "in.hvac_has_shared_system",
    "in.ashrae_iecc_climate_zone_2004_2_a_split",
    "in.misc_freezer",
    "in.misc_gas_fireplace",
    "in.radiant_barrier",
    "in.pv_orientation",
    "out.params.floor_area_conditioned_infiltration_reduction_ft_2_delta_ach_50",
    "in.window_areas",
    "in.geometry_stories",
    "in.mechanical_ventilation",
    "in.hvac_system_single_speed_ac_airflow",
    "out.params.slab_perimeter_exposed_conditioned_ft",
    "in.cooling_setpoint_offset_magnitude",
    "in.state",
    "in.utility_bill_propane_marginal_rates",
    "in.hvac_system_is_faulted",
    "in.weather_file_longitude",
    "in.location_region",
    "in.orientation",
    "in.ground_thermal_conductivity",
    "in.clothes_washer_usage_level",
    "in.simulation_control_run_period_begin_month",
    "in.natural_ventilation",
    "in.utility_bill_simple_filepaths",
    "in.dishwasher_usage_level",
    "in.simulation_control_run_period_end_day_of_month",
    "in.misc_pool",
    "in.insulation_roof",
    "in.income",
    "in.water_heater_in_unit",
    "in.hot_water_distribution",
    "in.plug_loads",
    "in.utility_bill_scenario_names",
    "in.vintage",
    "in.heating_setpoint_has_offset",
    "in.eaves",
    "in.refrigerator_usage_level",
    "in.geometry_attic_type",
    "in.generation_and_emissions_assessment_region",
    "in.hvac_secondary_heating_fuel",
    "out.params.door_area_ft_2",
    "in.geometry_story_bin",
    "in.pv_system_size",
    "in.hvac_system_single_speed_ashp_airflow",
    "in.lighting_other_use",
    "in.misc_well_pump",
    "in.infiltration",
    "in.refrigerator",
    "in.heating_setpoint",
    "in.census_division_recs",
    "in.water_heater_fuel",
    "in.simulation_control_run_period_calendar_year",
    "in.bedrooms",
    "in.hot_water_fixtures",
    "in.duct_leakage_and_insulation",
    "in.water_heater_location",
    "in.weather_file_latitude",
    "in.cooling_setpoint",
    "in.lighting",
    "in.solar_hot_water",
    "in.geometry_building_type_acs",
    "in.misc_pool_pump",
    "in.misc_gas_lighting",
    "in.aiannh_area",
    "in.geometry_garage",
    "out.params.size_water_heater_gal",
    "in.geometry_stories_low_rise",
    "out.params.rim_joist_area_above_grade_exterior_ft_2",
    "in.geometry_building_number_units_sfa",
    "in.hvac_secondary_heating_partial_space_conditioning",
    "in.simulation_control_run_period_begin_day_of_month",
    "in.county_name",
    "floor_area_m2",
    # is_missing__* features omitted for brevity; script handles them automatically if included
]

COM_FEATURES = [
    "in.comstock_building_type",
    "in.comstock_building_type_group",
    "in.building_subtype",
    "in.floor_area_category",
    "in.number_of_stories",
    "in.aspect_ratio",
    "in.wall_construction_type",
    "in.window_to_wall_ratio_category",
    "in.window_type",
    "in.vintage",
    "in.year_built",
    "in.building_america_climate_zone",
    "in.ashrae_iecc_climate_zone_2006",
    "in.state",
    "in.census_region_name",
    "in.census_division_name",
    "in.cambium_grid_region",
    "in.iso_rto_region",
    "in.cluster_id",
    "in.cluster_name",
    "in.county_name",
    "in.weather_file_2018",
    "in.weather_file_tmy3",
    "in.hvac_system_type",
    "in.hvac_category",
    "in.hvac_heat_type",
    "in.hvac_cool_type",
    "in.hvac_vent_type",
    "in.hvac_combined_type",
    "in.heating_fuel",
    "in.hvac_night_variability",
    "in.weekday_opening_time",
    "in.weekday_operating_hours",
    "in.weekend_opening_time",
    "in.weekend_operating_hours",
    "in.energy_code_followed_during_original_building_construction",
    "in.energy_code_followed_during_last_hvac_replacement",
    "in.energy_code_followed_during_last_roof_replacement",
    "in.energy_code_followed_during_last_walls_replacement",
    "in.ownership_type",
    "in.party_responsible_for_operation",
    "in.purchase_input_responsibility",
    "out.params.hdd65f",
    "out.params.hdd50f",
    "out.params.cdd65f",
    "out.params.average_window_shgc",
    "out.params.window_to_wall_ratio",
    "out.params.building_fraction_heated",
    "out.params.boiler_average_efficiency",
    "out.params.average_heating_setpoint_max",
    "out.params.average_heating_setpoint_min",
    "out.params.hours_heating_setpoint_not_met",
    "out.params.average_outdoor_air_fraction",
    "floor_area_ft2",
    "floor_area_m2",
    "year_built_num",
    "operating_hours_week",
    "heating_setpoint_span_c",
    # is_missing__* features omitted for brevity; script handles them automatically if included
]

def load_tsv(path, has_allowable):
    tsv = {}
    with open(path, newline="", encoding="utf-8") as f:
        reader = csv.DictReader(f, delimiter="\t")
        for row in reader:
            name = row.get("field_name")
            if not name:
                continue
            tsv[name] = {
                "location": row.get("field_location", ""),
                "data_type": row.get("data_type", ""),
                "units": row.get("units", ""),
                "description": row.get("field_description", ""),
                "allowable": row.get("allowable_baseline_enumerations" if has_allowable else "allowable_enumeration", ""),
            }
    return tsv

def guide_for_feature(name, meta):
    if name.startswith("is_missing__"):
        return "Engineered missing-flag: 1 if base feature missing, else 0."
    if name in {"floor_area_m2", "floor_area_ft2", "operating_hours_week", "year_built_num", "heating_setpoint_span_c"}:
        return "Engineered numeric: computed in app/training, not user-entered directly."
    if meta and meta.get("allowable"):
        return "Categorical: must match allowable enumerations in TSV or '__MISSING__'."
    return "Numeric: provide value in units from TSV; use NaN/blank to trigger missing handling."

def build_mapping(features, tsv):
    rows = []
    for f in features:
        meta = tsv.get(f)
        rows.append({
            "feature": f,
            "data_type": meta.get("data_type") if meta else "",
            "units": meta.get("units") if meta else "",
            "description": meta.get("description") if meta else "(engineered or not in TSV)",
            "allowable": meta.get("allowable") if meta else "",
            "user_input_guidance": guide_for_feature(f, meta),
        })
    return rows

def main():
    res_tsv = load_tsv(Path("USstock/data_dictionary.tsv"), has_allowable=True)
    com_tsv = load_tsv(Path("USstock/comstock/data_dictionary.tsv"), has_allowable=False)

    res_rows = build_mapping(RES_FEATURES, res_tsv)
    com_rows = build_mapping(COM_FEATURES, com_tsv)

    out = Path("feature_mapping.csv")
    with open(out, "w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(
            f,
            fieldnames=["model", "feature", "data_type", "units", "description", "allowable", "user_input_guidance"]
        )
        writer.writeheader()
        for r in res_rows:
            r["model"] = "ResStock"
            writer.writerow(r)
        for r in com_rows:
            r["model"] = "ComStock"
            writer.writerow(r)

    print(f"Wrote {out}")

if __name__ == "__main__":
    main()
```
